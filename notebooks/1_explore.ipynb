```python
text = df["jobdescription"].str.lower()
for kw in ["excel", "sql", "python", "communication", "java", "tally", "sales"]:
    print(kw, text.str.contains(kw, na=False).sum())
```

```python
skills = ["excel", "sql", "python", "communication", "tally", "sales"]
counts = {s: text.str.contains(s, na=False).sum() for s in skills}
pd.Series(counts).sort_values(ascending=False)
```

```python
skills = ["excel", "sql", "python", "communication", "tally", "sales",
          "accounting", "teaching", "customer service", "marketing"]
```

```python
import pandas as pd
df = pd.read_csv("../data/raw/naukri_com-job_sample.csv")
text = df["jobdescription"].str.lower()
counts = {s: text.str.contains(s, na=False).sum() for s in skills}
pd.Series(counts).sort_values(ascending=False)
```

```python
import re
clean = df["jobdescription"].str.lower().str.replace("send me jobs like this", "", regex=False)
counts = {s: clean.str.contains(rf"\b{re.escape(s)}\b", na=False).sum() for s in skills}
pd.Series(counts).sort_values(ascending=False)
```

```python
counts = {s: clean.str.contains(rf"\b{re.escape(s)}s?\b", na=False).sum() for s in skills}
pd.Series(counts).sort_values(ascending=False)
```
