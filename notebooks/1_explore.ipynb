```python
skills = ["excel", "sql", "python", "communication", "tally", "sales",
          "accounting", "teaching", "customer service", "marketing"]
```

```python
import pandas as pd
df = pd.read_csv("../data/raw/naukri_com-job_sample.csv")
text = df["jobdescription"].str.lower()
counts = {s: text.str.contains(s, na=False).sum() for s in skills}
pd.Series(counts).sort_values(ascending=False)
```

```python
import re
clean = df["jobdescription"].str.lower().str.replace("send me jobs like this", "", regex=False)
counts = {s: clean.str.contains(rf"\b{re.escape(s)}\b", na=False).sum() for s in skills}
pd.Series(counts).sort_values(ascending=False)
```

```python
counts = {s: clean.str.contains(rf"\b{re.escape(s)}s?\b", na=False).sum() for s in skills}
pd.Series(counts).sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

top = df["industry"].value_counts().head(10)
top.sort_values().plot(kind="barh")
plt.title("Top 10 industries by job postings")
plt.xlabel("Job postings")
plt.tight_layout()
plt.show()
```

```python
(df["industry"].value_counts(normalize=True).head(10) * 100).round(1)
```

```python
df["site_name"].value_counts()
df["postdate"].head()
```

```python
df["postdate"] = pd.to_datetime(df["postdate"], errors="coerce", utc=True)
print(df["postdate"].min(), df["postdate"].max())
print(df["postdate"].dt.year.value_counts().sort_index())
print(df["site_name"].value_counts())
```

```python
top_inds = df["industry"].value_counts().head(5).index
rows = []
for ind in top_inds:
    sub = clean[df["industry"] == ind]
    for s in skills:
        share = sub.str.contains(rf"\b{re.escape(s)}s?\b", na=False).mean() * 100
        rows.append({"industry": ind, "skill": s, "share": share})
pd.DataFrame(rows).pivot(index="skill", columns="industry", values="share").round(1)
```
